토끼와 여우처럼 먹는 쪽과 먹히는 쪽의 수가 번갈아 오르내리는 모습을 흉내 냈어요. 피식자 출생률이 높으면 토끼가 빨리 늘어요. 포식률은 여우가 토끼를 잡는 정도라서 크면 토끼가 빨리 줄어요. 포식자 사망률이 높으면 여우가 빨리 줄어요. 포식 효율은 잡아먹은 먹이가 새끼로 이어지는 정도예요. 처음 피식자 수가 많으면 포식자가 크게 늘 것 같아요. 점심 메뉴 번호는 상관없을 것 같아요, 골라내 봐요! 먹이가 많으면 포식자가 늘고, 포식자가 늘면 먹이가 줄고, 먹이가 줄면 포식자도 줄어요. 이 순환이 주기를 만들어요. 진짜로 잰 숫자가 아니라 컴퓨터로 흉내 낸 연습용 데이터예요.

**핵심 속성(변수)**
- **피식자 출생률** — 먹이가 되는 동물의 태어나는 속도예요. 클수록 빨리 늘어요
- **포식률** — 포식자가 먹이 동물을 잡아먹는 정도예요. 클수록 먹이가 빨리 줄어요
- **포식자 사망률** — 포식자가 죽는 속도예요. 클수록 포식자가 줄어요
- **포식 효율** — 잡아먹은 먹이가 포식자 번식으로 바뀌는 정도예요. 클수록 포식자가 빨리 늘어요
- **결과** — 포식자가 가장 많았을 때의 마릿수예요

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 처음 피식자 수(마리) — 처음 피식자 수를 나타내는 값이에요.
- 점심 메뉴 번호 — 실험하던 날 점심 메뉴에 붙인 번호예요.
- 사용한 노트북 번호 — 실험에 쓴 노트북에 붙인 번호예요.
- 관찰한 요일 — 자료를 기록한 요일을 숫자로 나타낸 값이에요.
- 측정 장소 번호 — 자료를 모은 장소에 붙인 번호예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["피식자 출생률", "포식률", "포식자 사망률", "포식 효율"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. 심화 — 분류형 데이터 둘러보기
같은 모형의 분류형 데이터예요. 결과 `결과(진동 주기 50 이하 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c161_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["피식자 출생률", "포식률", "포식자 사망률", "포식 효율"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(진동 주기 50 이하 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```
