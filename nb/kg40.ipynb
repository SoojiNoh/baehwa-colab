## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/menu.csv')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

d = df.dropna(subset=['Total Fat', 'Sugars', 'Protein']).copy()
z = StandardScaler().fit_transform(d[['Total Fat', 'Sugars', 'Protein']])
m = KMeans(n_clusters=4, random_state=42, n_init=10)
d['군집'] = m.fit_predict(z)
print(d.groupby('군집')[['Total Fat', 'Sugars', 'Protein']].mean().round(2))
plt.scatter(d['Total Fat'], d['Sugars'], c=d['군집'], cmap='tab10', alpha=0.6)
plt.xlabel('Total Fat'); plt.ylabel('Sugars'); plt.show()
print(d.groupby('군집')['Item'].apply(lambda s: list(s.head(3))))
````

### 3-1. 세 영양소를 표준화한다.

````python
from sklearn.preprocessing import StandardScaler # 척도 맞추기
features = ['Total Fat', 'Sugars', 'Protein'] # 세 영양소
d = df.dropna(subset=features).copy() # 입력 결측 제외
z = StandardScaler().______(d[features]) # 영양소 표준화
print(pd.DataFrame(z, columns=features).______()) # 변환값 확인
# ✅ 세 영양소의 표준화된 값이 앞부분 표로 보이면 맞아요.
````
