# 📈 공공 공간과 동네 운동 모임의 형성 — 데이터 분석 빈칸 실습

동네 공원에서 사람들이 모여 운동하는 모임을 흉내 낸 데이터예요. 공원이 눈에 잘 띄는 정도가 크면 찾기 쉬워요. 모일 만한 거점이 있으면 모이기 좋아요. 모인 사람이 보일 때의 끌림이 크면 따라 나가요. 정해진 시간에 모이는 정도가 크면 습관이 돼요. ‘동네 운동 시설 수’는 클수록 결과가 커질 수도 있어요(조금 관련). 결과와 상관없을 것 같은 것도 섞여 있어요 — 골라내는 게 숙제! 다른 사람이 운동하는 게 보이면 나도 하고 싶어져요. 진짜로 잰 숫자가 아니라 컴퓨터로 흉내 낸 연습용 데이터예요.

**핵심 속성(변수)**
- **공원이 눈에 잘 띄는 정도**(점) — 공원이 잘 보이는 정도(점)예요. 높을수록 사람들이 찾아와요.
- **모일 만한 거점**(점) — 모이기 좋은 장소가 있는 정도(점)예요. 높을수록 모임이 쉽게 생겨요.
- **모인 사람이 보일 때의 끌림**(점) — 이미 모인 사람을 보고 끌리는 정도(점)예요. 클수록 새 참여자가 늘어요.
- **정해진 시간에 모이는 정도**(점) — 정해진 시간에 모이는 규칙성(점)이에요. 높을수록 모임이 안정돼요.
- **결과** — 운동 모임에 참여한 사람의 수(명)예요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 운동복 색깔 번호 — 운동복 색을 번호로 나타낸 거예요.
- 동네 이름 글자 수(자) — 동네 이름의 글자 수예요.
- 동네 운동 시설 수(개) — 동네에 있는 운동 시설의 수예요.
- 관찰한 요일 번호 — 관찰한 요일을 번호로 나타낸 거예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["공원이 눈에 잘 띄는 정도", "모일 만한 거점", "모인 사람이 보일 때의 끌림", "정해진 시간에 모이는 정도"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. 심화 — 분류형 데이터 둘러보기

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c1045_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["공원이 눈에 잘 띄는 정도", "모일 만한 거점", "모인 사람이 보일 때의 끌림", "정해진 시간에 모이는 정도"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(모임이 커졌는지 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```
