## 1. 데이터 불러오기
📖 **오픈북** — 「#20 과일 채소 딥러닝」 숙제00 `pd.read_csv(url)`

```python
dc = pd.______("https://baehwa-ai-2026.web.app/dataproject/csv/c776_c.csv")      # 힌트: CSV 를 읽는 함수
dc = dc.dropna(subset=["틱마다 소통 횟수", "두 집단 사이 신뢰", "소통 가능한 범위", "관리자 수", "결과(지연이 큰지 여부)"])      # 이번에 쓰는 열에 빈칸이 있는 줄만 지우기
print(dc.shape)
# 이상치 거르기 — 머신러닝 노트 3번과 같은 IQR
for col in ["틱마다 소통 횟수", "두 집단 사이 신뢰", "소통 가능한 범위", "관리자 수"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
print('거른 뒤', dc.shape)
```

▶ 이렇게 나오면 맞아요 (숫자는 조금 달라도 괜찮아요)
