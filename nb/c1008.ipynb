# 📈 공개 동료 심사의 효과 — 데이터 분석 빈칸 실습

심사 과정을 공개하는 방식을 컴퓨터로 흉내 냈어요. 심사자 이름 공개 비율은 높으면 심사 태도가 달라질 수도 있어요. 믿기 어려운 심사 확률과 과학자 수와 심사 점수가 빗나가는 폭은 데이터로 확인해 봐요. 심사자 수는 많으면 질이 오를 수도 있어요. 글자 크기나 심사자 평균 나이처럼 결과와 상관없을 것 같은 것도 섞여 있어요 — 골라내는 게 숙제! 이름이 공개되면 말을 더 조심스럽게 하게 돼요. 진짜로 잰 숫자가 아니라 컴퓨터로 흉내 낸 연습용 데이터예요.

**핵심 속성(변수)**
- **심사자 이름 공개 비율**(%) — 평가자 이름을 밝히는 비율이에요. 클수록 공개 심사가 많아요.
- **믿기 어려운 심사 확률**(확률) — 평가가 엉터리일 확률이에요. 클수록 못 믿을 평가가 많아요.
- **과학자 수**(명) — 논문을 쓰고 평가하는 사람 수예요. 클수록 큰 학계예요.
- **심사 점수가 빗나가는 폭**(%) — 평가 점수가 실제와 달라지는 정도예요. 클수록 점수가 부정확해요.
- **결과** — 뽑힌 논문들의 평균 질을 점수로 나타내요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 심사자 수(명) — 논문당 심사자 수예요.
- 그래프 글자 크기(pt) — 그래프에 쓰는 글자의 크기예요.
- 그래프 색깔 번호(번) — 그래프에서 쓰는 색깔의 번호예요.
- 심사자 평균 나이(살) — 심사자의 평균 나이예요.
- 저널 수(종) — 저널의 수예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["심사자 이름 공개 비율", "믿기 어려운 심사 확률", "과학자 수", "심사 점수가 빗나가는 폭"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 7-1. 정규화 — 숫자 크기를 0~1 로 맞추기

## 9. 심화 — 분류형 데이터 둘러보기
같은 모형의 분류형 데이터예요. 결과 `결과(질 높은 편 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c1008_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["심사자 이름 공개 비율", "믿기 어려운 심사 확률", "과학자 수", "심사 점수가 빗나가는 폭"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(질 높은 편 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```
