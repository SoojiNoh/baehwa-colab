````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/penguins_size.csv')
import matplotlib.pyplot as plt
s = df['body_mass_g']
t = s.fillna(s.median())
print('결측 전후', s.isna().sum(), t.isna().sum())
print(pd.DataFrame({'원본':s.describe(), '대체':t.describe()}))
plt.hist(s.dropna(), bins=15, alpha=0.5, label='원본')
plt.hist(t, bins=15, alpha=0.5, label='대체')
plt.legend(); plt.show()
````
