## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/StudentsPerformance.csv')
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
df['읽기기준이상'] = (df['reading score'] >= 70).astype(int)
d = df.dropna(subset=['읽기기준이상'])
X = d[['math score', 'writing score']]; y = d['읽기기준이상']
a, b, c, e = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
m = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), LogisticRegression(max_iter=1000))
m.fit(a, c); p = m.predict(b)
print('정확도', accuracy_score(e, p))
print('범주 순서', m.classes_); print(confusion_matrix(e, p, labels=m.classes_))
````

## 🤖 머신러닝으로 예측·분류하기
회귀 — 숫자를 맞혀요. 평균 오차(MAE)와 설명력(R²)을 봐요.

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

````python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split   # 학습용·시험용 나누기
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/StudentsPerformance.csv')
X = df[['reading score', 'writing score']]                     # 넣을 열(특징)
y = df['math score']                     # 맞힐 정답(숫자)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)   # 70% 학습 · 30% 시험
model = LinearRegression()   # 선형 회귀: 직선(평면)으로 맞추기
model.fit(X_train, y_train)          # 학습
pred = model.predict(X_test)         # 시험용으로 예측
print('평균 오차(MAE)', round(mean_absolute_error(y_test, pred), 2))   # 평균적으로 얼마나 틀렸나
print('설명력(R²)', round(r2_score(y_test, pred), 3))                  # 1에 가까울수록 잘 맞음
plt.scatter(y_test, pred, alpha=0.4); plt.xlabel('실제 ' + 'math score'); plt.ylabel('예측')
plt.title('실제 대 예측 (대각선에 모일수록 잘 맞음)'); plt.show()
````
