## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/penguins_size.csv')
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

d = df.dropna(subset=['species'])
X = d[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm']]; y = d['species']
a, b, c, e = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
m = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), DecisionTreeClassifier(max_depth=1, random_state=42))
m.fit(a, c); p = m.predict(b)
print('정확도', accuracy_score(e, p))
print('범주 순서', m.classes_); print(confusion_matrix(e, p, labels=m.classes_))
````

### 3-1. 깊이 1의 결과를 기록한다.
깊이 1 트리의 훈련·검증 정확도를 기록해 복잡도 비교 기준을 만들어요.

💡 **힌트**: fit으로 깊이 1 트리를 학습하고 score로 정확도를 구해요.

````python
from sklearn.model_selection import train_test_split; from sklearn.pipeline import make_pipeline # 분할과 연결
from sklearn.impute import SimpleImputer; from sklearn.tree import DecisionTreeClassifier # 트리 도구
d = df.dropna(subset=['species']); X = d[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm']]; y = d['species'] # 입력과 정답
a, b, c, e = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42) # 훈련과 검증
m = make_pipeline(SimpleImputer(strategy='median'), DecisionTreeClassifier(max_depth=1, random_state=42)); m.______(a, c) # 얕은 트리
scores = {'1': [m.______(a, c), m.score(b, e)]} # 훈련과 검증 정확도
print(pd.DataFrame(scores, index=['훈련', '검증']).T) # 기준 결과
# ✅ 깊이 1의 훈련·검증 정확도가 보이면 맞아요.
````

### 3-2. 깊이 3·제한 없음으로 바꿔 본다.
자료 분할을 고정하고 깊이 3과 제한 없는 트리의 성능을 비교해요.

💡 **힌트**: fit은 같은 자료로 학습, score는 훈련·검증 정확도 계산이에요.

````python
for depth in [3, None]: # 깊이 두 가지 비교
    m = make_pipeline(SimpleImputer(strategy='median'), DecisionTreeClassifier(max_depth=depth, random_state=42)) # 깊이만 변경
    m.______(a, c) # 같은 훈련 자료
    scores[str(depth)] = [m.______(a, c), m.score(b, e)] # 같은 검증 자료
print(pd.DataFrame(scores, index=['훈련', '검증']).T) # 세 깊이 비교
# ✅ 깊이 1·3·None의 훈련·검증 정확도가 세 행으로 보이면 맞아요.
````

### 3-3. 복잡할수록 항상 좋은지 설명한다.
훈련과 검증의 차이를 보고 복잡한 모델이 항상 좋은지 판단해요.

💡 **힌트**: DataFrame으로 깊이별 정확도를 표로 모아요.

````python
comparison = pd.______(scores, index=['훈련', '검증']).T # 깊이별 결과
comparison['훈련과검증차이'] = comparison['훈련'] - comparison['검증'] # 일반화 차이
print(comparison) # 성능 차이 확인
print('훈련 정확도가 높아져도 검증 정확도가 함께 높아지는지 확인해요. 한 번의 분할만으로 최적 깊이를 확정하지 않아요.') # 복잡도 해석
# ✅ 세 깊이의 정확도와 훈련·검증 차이가 보이면 맞아요.
````
