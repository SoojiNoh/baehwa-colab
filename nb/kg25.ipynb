## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/winequality-red.csv')
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

d = df.dropna(subset=['quality'])
X = d[['alcohol', 'volatile acidity', 'sulphates']]; y = d['quality']
a, b, c, e = train_test_split(X, y, test_size=0.25, random_state=42)
m = make_pipeline(SimpleImputer(strategy='median'), LinearRegression())
m.fit(a, c); p = m.predict(b)
print('R²', r2_score(e, p), 'MAE', mean_absolute_error(e, p))
plt.scatter(e, p, alpha=0.4)
plt.plot([e.min(),e.max()], [e.min(),e.max()], 'r--')
plt.xlabel('실제'); plt.ylabel('예측'); plt.show()
````

## 🤖 머신러닝으로 예측·분류하기
분류 — 범주를 맞혀요. 시험 정확도와 혼동행렬을 봐요.

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

````python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split   # 학습용·시험용 나누기
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/winequality-red.csv')
df['좋은 와인(6점 이상)'] = (df['quality'] >= 6).map({True: '좋음', False: '보통'})
X = df[['alcohol', 'volatile acidity', 'sulphates', 'citric acid']]                     # 넣을 열(특징)
y = df['좋은 와인(6점 이상)']                     # 맞힐 정답(범주)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)   # 70% 학습 · 30% 시험
model = DecisionTreeClassifier(max_depth=3, random_state=0)   # 결정트리(질문 3단계까지)
model.fit(X_train, y_train)          # 학습
pred = model.predict(X_test)         # 시험용으로 예측
print('시험 정확도', round(accuracy_score(y_test, pred), 3))   # 맞힌 비율
ConfusionMatrixDisplay.from_predictions(y_test, pred, xticks_rotation=45)   # 혼동행렬: 무엇을 무엇으로 헷갈렸나
plt.title('혼동행렬'); plt.tight_layout(); plt.show()
````
