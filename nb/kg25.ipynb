## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/winequality-red.csv')
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

d = df.dropna(subset=['quality'])
X = d[['alcohol', 'volatile acidity', 'sulphates']]; y = d['quality']
a, b, c, e = train_test_split(X, y, test_size=0.25, random_state=42)
m = make_pipeline(SimpleImputer(strategy='median'), LinearRegression())
m.fit(a, c); p = m.predict(b)
print('R²', r2_score(e, p), 'MAE', mean_absolute_error(e, p))
plt.scatter(e, p, alpha=0.4)
plt.plot([e.min(),e.max()], [e.min(),e.max()], 'r--')
plt.xlabel('실제'); plt.ylabel('예측'); plt.show()
````
