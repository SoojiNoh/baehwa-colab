## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Pokemon.csv')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

d = df.dropna(subset=['Attack', 'Defense', 'Speed']).copy()
z = StandardScaler().fit_transform(d[['Attack', 'Defense', 'Speed']])
m = KMeans(n_clusters=4, random_state=42, n_init=10)
d['군집'] = m.fit_predict(z)
print(d.groupby('군집')[['Attack', 'Defense', 'Speed']].mean().round(2))
plt.scatter(d['Attack'], d['Defense'], c=d['군집'], cmap='tab10', alpha=0.6)
plt.xlabel('Attack'); plt.ylabel('Defense'); plt.show()
````

## 4. 내 생각 정리 (선택)
✍️ **(선택 — 필수 아니에요)** 답은 수행평가 사이트의 칸에 써요.

- 공격·방어·속도 조합에 따른 군집을 해석한다.
- 네 군집을 만든다.
- 평균값으로 특징을 설명한다.

- 이 결과를 그대로 믿기 어려운 이유 → 「1-3. 한계」
- 내 코드에서 중요한 3줄과 이유 → 「1-1. 이 코드 3줄은 왜 썼나」

## 🤖 머신러닝으로 예측·분류하기
분류 — 범주를 맞혀요. 시험 정확도와 혼동행렬을 봐요.

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

````python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split   # 학습용·시험용 나누기
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Pokemon.csv')
X = df[['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']]                     # 넣을 열(특징)
y = df['Legendary']                     # 맞힐 정답(범주)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)   # 70% 학습 · 30% 시험
model = DecisionTreeClassifier(max_depth=3, random_state=0)   # 결정트리(질문 3단계까지)
model.fit(X_train, y_train)          # 학습
pred = model.predict(X_test)         # 시험용으로 예측
print('시험 정확도', round(accuracy_score(y_test, pred), 3))   # 맞힌 비율
ConfusionMatrixDisplay.from_predictions(y_test, pred, xticks_rotation=45)   # 혼동행렬: 무엇을 무엇으로 헷갈렸나
plt.title('혼동행렬'); plt.tight_layout(); plt.show()
````
