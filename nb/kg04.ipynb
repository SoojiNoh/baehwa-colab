카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/vgsales.csv')
import matplotlib.pyplot as plt
g = df.groupby('Genre')['Global_Sales'].agg(['sum','count'])
print(g.sort_values('sum', ascending=False))
g['sum'].sort_values().plot.barh()
plt.xlabel('세계 판매량(백만 장)'); plt.show()
````
