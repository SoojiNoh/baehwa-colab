학교와 직장을 막으면 감염이 줄어드는지 컴퓨터로 흉내 냈어요. 휴교 정도와 재택근무 정도와 거리 두기 정도는 높으면 정점이 낮아질 수도 있어요. 평균 접촉자 수와 회복 확률과 잠복 기간은 데이터로 확인해 봐요. 마스크 착용 비율과 백신 접종률은 높으면 정점이 낮아질 수도 있어요. 이름 글자 수나 색깔 번호처럼 결과와 상관없을 것 같은 것도 섞여 있어요 — 골라내는 게 숙제! 사람이 모이는 곳이 줄면 병이 퍼지는 길도 줄어요. 진짜로 잰 숫자가 아니라 컴퓨터로 흉내 낸 연습용 데이터예요.

**핵심 속성(변수)**
- **휴교 정도**(%) — 학교를 닫는 정도예요. 클수록 학교 접촉이 많이 끊겨요.
- **재택근무 정도**(%) — 집에서 일하는 정도예요. 클수록 직장 접촉이 줄어요.
- **거리 두기 정도**(%) — 서로 떨어져 지내는 정도예요. 클수록 접촉이 줄어요.
- **평균 접촉자 수**(명) — 한 사람이 만나는 사람 수예요. 클수록 병이 잘 옮아요.
- **회복 확률**(%) — 아픈 사람이 나을 확률이에요. 클수록 빨리 나아요.
- **잠복 기간**(일) — 감염되고 증상이 나타나기까지 걸리는 날수예요. 클수록 오래 모르고 지내요.
- **결과** — 동시에 아픈 사람이 가장 많을 때의 비율이에요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 마스크 착용 비율(%) — 마스크 쓰는 비율이에요.
- 백신 접종률(%) — 백신을 맞은 비율이에요.
- 조사자 이름 글자 수(글자) — 조사한 사람 이름의 글자 수예요.
- 그래프 글자 크기(pt) — 그래프에 쓰는 글자의 크기예요.
- 그래프 색깔 번호(번) — 그래프에서 쓰는 색깔의 번호예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["휴교 정도", "재택근무 정도", "거리 두기 정도", "평균 접촉자 수", "회복 확률", "잠복 기간"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. 심화 — 분류형 데이터 둘러보기
같은 모형의 분류형 데이터예요. 결과 `결과(정점이 높은지 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c1005_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["휴교 정도", "재택근무 정도", "거리 두기 정도", "평균 접촉자 수", "회복 확률", "잠복 기간"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(정점이 높은지 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```

▶ 이렇게 나오면 맞아요 (숫자는 조금 달라도 괜찮아요)



## 9-1. 분류 모델 만들고 평가하기 — confusion matrix
