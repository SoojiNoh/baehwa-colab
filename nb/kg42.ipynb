## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Iris.csv')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

d = df.dropna(subset=['PetalLengthCm', 'PetalWidthCm']).copy()
z = StandardScaler().fit_transform(d[['PetalLengthCm', 'PetalWidthCm']])
m = KMeans(n_clusters=3, random_state=42, n_init=10)
d['군집'] = m.fit_predict(z)
print(d.groupby('군집')[['PetalLengthCm', 'PetalWidthCm']].mean().round(2))
plt.scatter(d['PetalLengthCm'], d['PetalWidthCm'], c=d['군집'], cmap='tab10', alpha=0.6)
plt.xlabel('PetalLengthCm'); plt.ylabel('PetalWidthCm'); plt.show()
print(pd.crosstab(d['군집'], d['Species']))
````
