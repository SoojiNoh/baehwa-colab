# 📈 죄수의 딜레마와 전략 대결 — 데이터 분석 빈칸 실습

죄수의 딜레마는 서로 협력하면 좋지만 혼자 배신하면 이득인 상황을 흉내 낸 연습용 데이터예요. 한 번만 만나면 배신이 이득이지만 계속 만나면 상대의 행동을 따라 하는 전략이 유리해요. 만나는 횟수가 늘면 협력 전략이 더 앞서는지 알아보면 재미있어요.

**핵심 속성(변수)**
- **반복해서 만난 횟수**(회) — 같은 상대를 다시 만나는 횟수예요. 많을수록 협력이 유리해져요
- **배신했을 때 얻는 점수**(점) — 배신했을 때 혼자 얻는 점수예요. 클수록 배신하고 싶어져요
- **상대의 협력 비율**(%) — 상대가 협력하는 비율이에요. 높을수록 협력이 쉬워요
- **참여자 수**(명) — 게임에 참여한 사람의 수예요. 많을수록 다양한 전략이 만나요
- **결과** — 조건부 협력 전략의 평균 점수에서 배신 전략의 평균 점수를 뺀 값이에요

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 설문지 인쇄 색 번호(번) — 설문지 인쇄 색에 붙인 번호예요.
- 화면 배경 색 번호(번) — 프로그램 화면 배경에 쓴 색을 나타내는 번호예요.
- 기록한 사람 이름 글자 수(자) — 자료를 기록한 사람 이름이 몇 글자인지 센 값이에요.
- 상대를 기억하는 횟수(회) — 이전 만남을 몇 번까지 기억하는지 나타내요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["반복해서 만난 횟수", "배신했을 때 얻는 점수", "상대의 협력 비율", "참여자 수"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. (더 해 보기) 분류형 데이터
같은 모형의 분류형 데이터예요. 결과 `결과(협력 전략이 앞서는지 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c664_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["반복해서 만난 횟수", "배신했을 때 얻는 점수", "상대의 협력 비율", "참여자 수"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(협력 전략이 앞서는지 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```

▶ 이렇게 나오면 맞아요 (숫자는 조금 달라도 괜찮아요)
