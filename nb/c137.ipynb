# 📈 등온 압축과 기체 압력 — 데이터 분석 빈칸 실습

상자 속 알갱이를 눌러 압축하면 압력이 어떻게 달라지는지 컴퓨터로 흉내 냈어요. 입자 수가 많으면 부딪힘이 늘어 압력이 커질 것 같아요. 처음 속력이 빠르면 더 세게 부딪혀 압력이 오를 수도 있어요. 입자 질량이 크면 부딪힐 때 힘이 더 클 수도 있고, 상자 높이는 부피를 정해요. 이것들이 평균 압력에 어떻게 이어지는지는 데이터로 확인해 봐요. 기체 온도는 조금 관련 있을 수도 있고, 상자 폭이 크면 압력이 줄 수도 있어요. 색깔 번호, 입자 크기, 벽 두께는 상관없을 것 같으니 골라내는 게 숙제! 공기 펌프를 눌러 본 적 있다면 감이 올 거예요. 진짜로 잰 숫자가 아니라 컴퓨터로 흉내 낸 연습용 데이터예요.

**핵심 속성(변수)**
- **입자 수**(개) — 상자 속 알갱이의 개수예요. 클수록 많이 부딪혀요.
- **처음 속력** — 알갱이가 처음 움직이는 빠르기예요. 클수록 세게 부딪혀요.
- **입자 질량** — 알갱이 하나의 무게예요. 클수록 묵직하게 부딪혀요.
- **상자 높이**(%) — 상자의 세로 길이 비율이에요. 클수록 공간이 넓어요.
- **결과** — 상자 안 평균 압력을 나타내요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 기체 온도(K) — 온도예요.
- 그래프 색깔 번호(번) — 그래프에서 쓰는 색깔의 번호예요.
- 상자 폭(칸) — 폭이에요.
- 입자 크기(단위) — 크기예요.
- 벽 두께(칸) — 벽 두께예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["입자 수", "처음 속력", "입자 질량", "상자 높이"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 7-1. 정규화 — 숫자 크기를 0~1 로 맞추기
속성마다 숫자 크기가 달라요(예: 나이 15 와 소득 3,000,000). **가장 작은 값은 0, 가장 큰 값은 1** 이 되게 바꿔요.

기준(최솟값·최댓값)은 **학습용에서만** 구해요 — 시험지를 미리 보면 안 되니까요. 시험용도 같은 기준으로 바꿔요.

선형 회귀는 정규화해도 점수가 거의 같지만, 딥러닝(신경망)은 정규화를 안 하면 잘 못 배워요. 습관처럼 해 둬요.

## 9. (더 해 보기) 분류형 데이터
같은 모형의 분류형 데이터예요. 결과 `결과(압력 기준치 초과 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c137_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["입자 수", "처음 속력", "입자 질량", "상자 높이"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(압력 기준치 초과 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```
