## 🤖 머신러닝으로 예측·분류하기
회귀 — 숫자를 맞혀요. 평균 오차(MAE)와 설명력(R²)을 봐요.

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

````python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/AirPassengers.csv')
df['순서'] = range(len(df))                       # 몇 번째 달인가
df['월'] = pd.to_datetime(df['Month']).dt.month   # 몇 월인가
X = pd.get_dummies(df[['순서', '월']], columns=['월'])   # 추세 + 달마다 다른 계절성
y = df['#Passengers']
X_train, X_test, y_train, y_test = X[:-24], X[-24:], y[:-24], y[-24:]   # 마지막 2년을 시험용(미래)
model = LinearRegression()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('평균 오차(MAE)', round(mean_absolute_error(y_test, pred), 1)); print('설명력(R²)', round(r2_score(y_test, pred), 3))
plt.plot(df['Month'], y, label='실제'); plt.plot(df['Month'].iloc[-24:], pred, label='예측'); plt.xticks(df['Month'][::24]); plt.legend(); plt.show()
````
