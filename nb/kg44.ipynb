카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

```python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Mall_Customers.csv')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
z = StandardScaler().fit_transform(df[['Annual Income (k$)','Spending Score (1-100)']])
ks = range(1, 9)
values = [KMeans(n_clusters=k, random_state=42, n_init=10).fit(z).inertia_ for k in ks]
plt.plot(ks, values, marker='o'); plt.xlabel('군집 수'); plt.ylabel('관성'); plt.show()
```

## 🤖 머신러닝으로 예측·분류하기
군집 — 정답 없이 비슷한 것끼리 묶어요(K-평균).

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

```python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Mall_Customers.csv')
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
model = KMeans(n_clusters=5, n_init=10, random_state=0)   # 정답 없이 비슷한 손님끼리 5묶음
df['묶음'] = model.fit_predict(X)
print(df.groupby('묶음')[['Annual Income (k$)', 'Spending Score (1-100)']].mean().round(1))   # 묶음마다 평균
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=df['묶음'], cmap='tab10'); plt.xlabel('Annual Income (k$)'); plt.ylabel('Spending Score (1-100)'); plt.title('K-평균 군집'); plt.show()
```
