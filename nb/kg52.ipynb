## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/mnist_train_6000.csv', nrows=5000)
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
X = df.iloc[:,1:].astype('float32') / 255
y = df.iloc[:,0]
a, b, c, e = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
m = KNeighborsClassifier(n_neighbors=3)
m.fit(a, c); p = m.predict(b)
print('정확도', accuracy_score(e, p)); print(confusion_matrix(e, p, labels=range(10)))
````
