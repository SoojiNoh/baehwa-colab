# 🎬 한국어 영화 리뷰 감정 분류 (심화) — 딥러닝

캐글·NSMC 예제에서 많이 쓰는 임베딩 신경망 흐름을 줄였어요(글자 단위).

- (심화) 노트예요: 빈칸 없이 끝까지 돌아가요. 한 칸씩 실행하며 숫자와 그래프를 읽고, 보고서 「1-1·1-2·1-3」에 써요.

```python
# 0. 준비 — 그대로 실행해요
!pip -q install koreanize-matplotlib   # 한글 그래프 글꼴
import koreanize_matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. 데이터 불러오기

```python
df = pd.read_csv('https://raw.githubusercontent.com/e9t/nsmc/master/ratings_train.txt', sep='\t').dropna().sample(n=20000, random_state=0)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['document'].values, df['label'].values, test_size=0.2, random_state=0)
print(len(X_train))
```

## 2. 글자 → 번호, 신경망

```python
import keras
from keras import layers
keras.utils.set_random_seed(0)
vec = layers.TextVectorization(max_tokens=4000, output_sequence_length=60, split='character')
vec.adapt(X_train)
model = keras.Sequential([keras.Input(shape=(1,), dtype='string'), vec, layers.Embedding(4000, 32), layers.Conv1D(64, 3, activation='relu'), layers.GlobalMaxPooling1D(), layers.Dense(1, activation='sigmoid')])
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
h = model.fit(X_train, y_train, epochs=4, batch_size=128, validation_split=0.1, verbose=0)
plt.plot(h.history['accuracy'], label='학습'); plt.plot(h.history['val_accuracy'], label='검증'); plt.legend(); plt.title('학습 곡선'); plt.show()
```

## 3. 성능평가

```python
from sklearn.metrics import accuracy_score, f1_score
p = (model.predict(X_test, verbose=0)[:, 0] > 0.5).astype(int)
print('정확도', round(accuracy_score(y_test, p), 3), '· F1', round(f1_score(y_test, p), 3))
```
