### 3-1. 강수량의 0 비율을 구한다.

💡 **힌트**: eq로 0과 비교하고 mean으로 비율을 구해요.

### 3-2. IQR 상한과 후보 수를 계산한다.
IQR 상한과 후보 수를 구해 0이 많은 자료에서 기준이 어떻게 작동하는지 확인해요.

💡 **힌트**: quantile로 사분위수를 구하고 gt로 상한 초과를 표시해요.

````python
rain = df['Rainfall(mm)']  # 강수량
q1, q3 = rain.______([0.25, 0.75])  # 사분위수
upper = q3 + 1.5 * (q3 - q1)  # IQR 상한
rain_candidates = rain.______(upper)  # 상한 초과
print('IQR:', q3 - q1, '상한:', upper, '후보 수:', rain_candidates.sum())
# ✅ IQR·상한·후보 개수가 보이면 맞아요.
````

### 3-3. 비가 온 시간을 무조건 삭제하면 생길 문제를 쓴다.
