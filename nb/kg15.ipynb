카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/SeoulBikeData.csv', encoding='latin-1')
s = df['Rainfall(mm)']
q1, q3 = s.quantile([0.25, 0.75]); hi = q3+1.5*(q3-q1)
print('0 비율', s.eq(0).mean(), '상한', hi)
print('상한 초과', s.gt(hi).sum())
print(df.loc[s.gt(hi), ['Rainfall(mm)','Rented Bike Count']].describe())
````
