## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/menu.csv')
s = df['Calories']
q1, q3 = s.quantile([0.25, 0.75]); iqr = q3-q1
mask = (s < q1-1.5*iqr) | (s > q3+1.5*iqr)
print('후보 수', int(mask.sum()))
print(df.loc[mask, ['Item','Serving Size','Calories']])
````

## 3. 한 단계씩 해 보기
카드의 탐구 단계를 셀 하나씩 따라가요. `______` 칸에 함수 이름을 채우고 실행해요.

````python
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/menu.csv')   # 시작 코드가 바꿔 둔 표를 처음 상태로 다시 불러와요 — 그대로 실행
````

### 3-1. 열량의 사분위수를 구한다.
열량의 가운데 50% 구간을 정할 사분위수를 구해요.

💡 **힌트**: 지정한 비율의 경곗값은 quantile로 구해요.

````python
q1, q3 = df['Calories'].______([0.25, 0.75])  # 사분위수
print('1사분위수:', q1, '3사분위수:', q3)
# ✅ 열량의 1사분위수와 3사분위수가 보이면 맞아요.
````

### 3-2. IQR 경계와 후보 메뉴를 찾는다.
IQR 경계를 벗어난 메뉴를 삭제 후보로 살펴봐요.

💡 **힌트**: 후보 메뉴를 열량순으로 정렬하는 함수는 sort_values예요.

````python
iqr = q3 - q1  # 사분위 범위
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr  # 후보 경계
outliers = df.loc[(df['Calories'] < lower) | (df['Calories'] > upper)]  # 경계 밖 메뉴
print('하한:', lower, '상한:', upper, '후보 수:', len(outliers))
print(outliers[['Item', 'Calories']].______('Calories', ascending=False))
# ✅ 하한·상한·후보 수와 경계 밖 메뉴 목록이 보이면 맞아요.
````

### 3-3. 제공량을 보고 삭제 여부를 판단한다.
후보 메뉴의 제공량을 보고 실제 특성인지 오류인지 판단해요.

💡 **힌트**: 제공량과 함께 열량순으로 보는 함수는 sort_values예요.

````python
print(outliers[['Item', 'Serving Size', 'Calories']].______('Calories', ascending=False))  # 제공량 확인
print('제공량이 크거나 실제 고열량 메뉴일 수 있어요. 단위와 입력 오류를 확인한 뒤 목적에 맞게 삭제 여부를 판단해요.')
# ✅ 후보 메뉴의 제공량과 열량, 삭제 판단 안내가 보이면 맞아요.
````
