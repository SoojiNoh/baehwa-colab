## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/SeoulBikeData.csv', encoding='latin-1')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

d = df.dropna(subset=['Temperature(°C)', 'Humidity(%)', 'Wind speed (m/s)']).copy()
z = StandardScaler().fit_transform(d[['Temperature(°C)', 'Humidity(%)', 'Wind speed (m/s)']])
m = KMeans(n_clusters=4, random_state=42, n_init=10)
d['군집'] = m.fit_predict(z)
print(d.groupby('군집')[['Temperature(°C)', 'Humidity(%)', 'Wind speed (m/s)']].mean().round(2))
plt.scatter(d['Temperature(°C)'], d['Humidity(%)'], c=d['군집'], cmap='tab10', alpha=0.6)
plt.xlabel('Temperature(°C)'); plt.ylabel('Humidity(%)'); plt.show()
print(d.groupby('군집')['Rented Bike Count'].mean())
````
