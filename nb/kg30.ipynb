## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Pokemon.csv')
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

d = df.dropna(subset=['Legendary'])
X = d[['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']]; y = d['Legendary']
a, b, c, e = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
m = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), DecisionTreeClassifier(max_depth=4, class_weight='balanced', random_state=42))
m.fit(a, c); p = m.predict(b)
print('정확도', accuracy_score(e, p))
print('범주 순서', m.classes_); print(confusion_matrix(e, p, labels=m.classes_))
````
