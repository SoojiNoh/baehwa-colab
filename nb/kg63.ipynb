## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/adult.csv')
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
cols = df.select_dtypes(include='object').columns
df[cols] = df[cols].apply(lambda s: s.str.strip())
y = df['income'].str.rstrip('.').eq('>50K').astype(int)
X = df[['age','education.num','hours.per.week']]
a, b, c, e = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)
m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
m.fit(a, c); p = m.predict(b)
z = pd.DataFrame({'집단':df.loc[b.index,'sex'], '실제':e, '예측':p})
z['정확'] = z['실제'].eq(z['예측'])
print(z.groupby('집단').agg(표본수=('실제','size'), 정확도=('정확','mean'), 양성예측률=('예측','mean')))
print('실제 양성 중 탐지율', z[z['실제'].eq(1)].groupby('집단')['예측'].agg(['count','mean']))
````
