## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/spam.csv', encoding='latin-1')
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix
d = df[['v1','v2']].dropna().drop_duplicates('v2')
a, b, c, e = train_test_split(d['v2'], d['v1'], test_size=0.25, stratify=d['v1'], random_state=42)
m = make_pipeline(TfidfVectorizer(max_features=5000), LogisticRegression(max_iter=1000, class_weight='balanced'))
m.fit(a, c); p = m.predict(b)
print('정확도', accuracy_score(e, p))
print('순서: ham, spam'); print(confusion_matrix(e, p, labels=['ham','spam']))
````
