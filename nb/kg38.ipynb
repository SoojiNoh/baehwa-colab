## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/Mall_Customers.csv')
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

d = df.dropna(subset=['Annual Income (k$)', 'Spending Score (1-100)']).copy()
z = StandardScaler().fit_transform(d[['Annual Income (k$)', 'Spending Score (1-100)']])
m = KMeans(n_clusters=5, random_state=42, n_init=10)
d['군집'] = m.fit_predict(z)
print(d.groupby('군집')[['Annual Income (k$)', 'Spending Score (1-100)']].mean().round(2))
plt.scatter(d['Annual Income (k$)'], d['Spending Score (1-100)'], c=d['군집'], cmap='tab10', alpha=0.6)
plt.xlabel('Annual Income (k$)'); plt.ylabel('Spending Score (1-100)'); plt.show()
````
