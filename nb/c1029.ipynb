# 📈 가축 가뭄 보험과 유목민 — 데이터 분석 빈칸 실습

유목민에게 가축은 재산이에요. 가뭄이 오면 가축이 굶어 죽어 큰 손해를 입어요. 이 데이터는 가뭄 보험에 든 가구가 얼마나 빨리 가축 수를 되찾는지 흉내 낸 연습용 합성 데이터예요. 하지만 가축을 덜 팔면 풀이 모자랄 수도 있어요. 보험이 정말 좋기만 한지 알아보면 재미있어요.

**핵심 속성(변수)**
- **보험에 든 가구 비율**(%) — 가뭄 보험에 가입한 집의 비율이에요. 클수록 보호받는 집이 많아요.
- **평균 강수량**(mm/년) — 한 해에 내리는 비의 양이에요. 클수록 풀이 잘 자라요.
- **강수량 흔들림**(mm/년) — 비 오는 양이 해마다 달라지는 폭이에요. 클수록 가뭄 위험이 커요.
- **가축 증가율**(비율) — 가축이 늘어나는 빠르기예요. 클수록 빨리 불어나요.
- **유목 가구 수**(가구) — 가축을 키우며 이동하는 집의 수예요. 클수록 풀밭을 나눠 써요.
- **결과** — 가뭄 뒤 가축이 한 해에 몇 마리 회복되는지 나타내요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 가구당 가축 수(마리) — 한 가구 가축 수예요.
- 그래프 글자 크기(pt) — 그래프에 쓰는 글자의 크기예요.
- 풀밭 면적(ha) — 풀밭 넓이예요.
- 시장 가축 가격(만원) — 가축 값이에요.
- 가구원 수(명) — 식구 수예요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["보험에 든 가구 비율", "평균 강수량", "강수량 흔들림", "가축 증가율", "유목 가구 수"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. (더 해 보기) 분류형 데이터
같은 모형의 분류형 데이터예요. 결과 `결과(빨리 회복하는지 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c1029_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["보험에 든 가구 비율", "평균 강수량", "강수량 흔들림", "가축 증가율", "유목 가구 수"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(빨리 회복하는지 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```

▶ 이렇게 나오면 맞아요 (숫자는 조금 달라도 괜찮아요)
