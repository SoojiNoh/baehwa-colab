## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/e9t/nsmc/master/ratings_train.txt', sep='\t')
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
d = df.dropna(subset=['document','label']).drop_duplicates('document').sample(n=10000, random_state=42)
a, b, c, e = train_test_split(d['document'], d['label'], test_size=0.25, stratify=d['label'], random_state=42)
m = make_pipeline(TfidfVectorizer(analyzer='char', ngram_range=(2,4), max_features=12000), LogisticRegression(max_iter=1000))
m.fit(a, c); p = m.predict(b)
z = pd.DataFrame({'글자수':b.str.len(), '정확':p == e})
z['집단'] = z['글자수'].le(10).map({True:'10자 이하', False:'10자 초과'})
print(z.groupby('집단')['정확'].agg(['count','mean']))
````
