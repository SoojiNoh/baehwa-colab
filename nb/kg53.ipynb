## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/fashion-mnist_train_6000.csv', nrows=6000)
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
tf.keras.utils.set_random_seed(42)
X = df.iloc[:,1:].to_numpy(dtype='float32') / 255
y = df['label'].to_numpy()
a, b, c, e = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
m = tf.keras.Sequential([tf.keras.layers.Input((784,)), tf.keras.layers.Dense(64, activation='relu'), tf.keras.layers.Dense(10, activation='softmax')])
m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
m.fit(a, c, epochs=5, batch_size=64, validation_split=0.2, verbose=0)
print('평가 손실·정확도', m.evaluate(b, e, verbose=0))
p = m.predict(b, verbose=0).argmax(axis=1)
print(confusion_matrix(e, p, labels=range(10)))
````
