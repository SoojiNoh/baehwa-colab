# 📈 학교 배정 방식과 학생의 지원 전략 — 데이터 분석 빈칸 실습

고입이나 대입처럼 학교를 지원할 때는 합격 가능성을 보고 지원 순서를 정해요. 이 모형은 학생이 점수를 받고 여러 학교 중 정원이 찬 학교는 못 가고 다음 지망으로 넘어가는 모습을 흉내 낸 연습용 합성 데이터예요. 정원과 지원 개수를 바꾸면 1지망에 들어가는 학생이 얼마나 되는지 알아보면 재미있어요.

**핵심 속성(변수)**
- **학생 수**(명) — 지원하는 학생의 수예요. 많을수록 경쟁이 치열해요.
- **학교별 정원**(명) — 한 학교가 받을 수 있는 학생 수예요. 많을수록 합격하기 쉬워요.
- **1단계 지원 개수**(개) — 처음 단계에서 지원할 수 있는 학교 수예요. 많을수록 선택지가 넓어요.
- **점수 표준편차**(점) — 학생 점수가 평균에서 흩어진 정도예요. 클수록 점수 차이가 커요.
- **결과** — 1지망 학교에 배정받은 학생의 비율(%)이에요.

**기타 속성** — 결과와 관련이 있을까? 상관을 보고 넣을지 뺄지 **내가 골라요**.
- 학교 건물 층수(층) — 학교 건물의 층 수예요.
- 학교 급식 만족도(점) — 급식에 대한 만족도예요.
- 통학 시간(분) — 집에서 학교까지 가는 시간이에요.
- 학교 이름 글자 수(자) — 학교 이름의 글자 수예요.
- 경쟁률 높은 학교 비율(%) — 지원이 몰리는 학교의 비율이에요.

## 3. 이상치(IQR) 처리
값을 작은 것부터 줄 세웠을 때 4분의 1 지점이 Q1, 4분의 3 지점이 Q3 예요. 둘 사이 거리를 IQR 이라고 해요(가운데 절반이 퍼진 폭). **Q1 − 1.5×IQR 보다 작거나 Q3 + 1.5×IQR 보다 크면 이상치**로 보고 지워요.

단, 1~4 등급이나 「거의 0인 횟수」처럼 **값 종류가 몇 개뿐인 열은 건너뛰어요** — 그런 열의 드문 값은 이상치가 아니라 정상이에요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제14 `quantile(0.25)` · 숙제15 `Q3 + 1.5 * IQR`

```python
for col in ["학생 수", "학교별 정원", "1단계 지원 개수", "점수 표준편차"]:
    q1 = df[col].______(0.25)        # 힌트: 분위수
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    if iqr == 0 or df[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    하한 = q1 - 1.5 * iqr
    상한 = q3 + ______ * iqr          # 힌트: 몇 배?
    df = df[(df[col] >= 하한) & (df[col] <= 상한)]
print('이상치 지운 뒤', df.shape)
```

## 9. (더 해 보기) 분류형 데이터
같은 모형의 분류형 데이터예요. 결과 `결과(1지망 배정 절반 넘김 여부)` 별로 속성 평균이 어떻게 다른지 묶어서 봐요.

📖 **오픈북** — 「#14 Iris 데이터 전처리(이상치)」 숙제04 `df.groupby('종류')`

```python
dc = pd.read_csv("https://baehwa-ai-2026.web.app/dataproject/csv/c274_c.csv").dropna()
# 이상치가 섞이면 평균이 튀어요 — 3번에서 배운 IQR 거르기를 여기서도 해요
for col in ["학생 수", "학교별 정원", "1단계 지원 개수", "점수 표준편차"]:
    q1, q3 = dc[col].quantile(0.25), dc[col].quantile(0.75)
    if q3 == q1 or dc[col].nunique() <= 5: continue      # 값 종류가 적은 열은 건너뛰기
    dc = dc[(dc[col] >= q1 - 1.5 * (q3 - q1)) & (dc[col] <= q3 + 1.5 * (q3 - q1))]
dc.______("결과(1지망 배정 절반 넘김 여부)").mean(numeric_only=True)   # 힌트: 묶어서(group) 보기
```

▶ 이렇게 나오면 맞아요 (숫자는 조금 달라도 괜찮아요)
