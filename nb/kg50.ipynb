## 2. 시작 코드 — 그대로 실행해 보기
카드의 시작 코드예요. 실행하고 결과를 아래 그림과 비교해요.

````python
import pandas as pd
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/spam.csv', encoding='latin-1')
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
d = df[['v1','v2']].dropna().drop_duplicates('v2').copy()
v = TfidfVectorizer(stop_words='english', max_features=3000)
x = v.fit_transform(d['v2'])
m = KMeans(n_clusters=4, random_state=42, n_init=10)
d['군집'] = m.fit_predict(x)
terms = v.get_feature_names_out()
for i, center in enumerate(m.cluster_centers_):
    print(i, terms[center.argsort()[-8:][::-1]])
print(pd.crosstab(d['군집'], d['v1'], normalize='index'))
````

- 내 코드에서 중요한 3줄과 이유 → 「1-1. 이 코드 3줄은 왜 썼나」

## 🤖 머신러닝으로 예측·분류하기
분류 — 범주를 맞혀요. 시험 정확도와 혼동행렬을 봐요.

- 학습용 70%로 배우고, 시험용 30%로 채점해요(`train_test_split`).
- 결과 숫자는 아래에 저장돼 있어요. 직접 실행해서 같은 숫자가 나오는지 확인해요.
- 보고서: 중요한 코드 3줄 → 「1-1」, 숫자가 뜻하는 것 → 「1-2」, 숫자가 낮은 이유 → 「1-3 한계」

````python
# 🤖 머신러닝 단계 — 그대로 실행해요
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split   # 학습용·시험용 나누기
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
df = pd.read_csv('https://baehwa-ai-2026.web.app/dataproject/csv/kaggle/spam.csv', encoding='latin-1')
df = df.dropna(subset=['v2']).sample(n=min(len(df), 5000), random_state=0)   # 5,000개만 — 빠르게
X_train, X_test, y_train, y_test = train_test_split(df['v2'].astype(str), df['v1'], test_size=0.3, random_state=0)
vec = TfidfVectorizer(max_features=3000)              # 글 → 낱말 점수표
Xtr = vec.fit_transform(X_train); Xte = vec.transform(X_test)
model = LogisticRegression(max_iter=1000)             # 로지스틱 회귀 분류
model.fit(Xtr, y_train)
pred = model.predict(Xte)
print('시험 정확도', round(accuracy_score(y_test, pred), 3))
ConfusionMatrixDisplay.from_predictions(y_test, pred)
plt.title('혼동행렬'); plt.show()
````
